```python
# Import necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV file into a Pandas DataFrame
file_path = 'Results/Results.csv'  # Replace with your file path

# Read the CSV file with semicolon as the delimiter
df = pd.read_csv(file_path, sep=';', quotechar='"')

# Display the first few rows of the DataFrame
df.head()

# Plot two columns using Seaborn
# Replace 'execution_time' and 'resources_used' with the actual column names you want to plot
plt.figure(figsize=(40, 6))
sns.scatterplot(x='file_name', y='execution_time', data=df)

# Add title and labels
plt.title('Scatter Plot of Execution Time vs Resources Used')
plt.xlabel('Execution Time')
plt.ylabel('File Name')

# Show the plot
plt.show()
```

```python
# Filter out rows where 'file_name' contains '3_5_15'
filtered_df = df[df['file_name'].str.contains('3_5_15')]

# Plot two columns using Seaborn
# Replace 'file_name' and 'execution_time' with the actual column names you want to plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='file_name', y='execution_time', data=filtered_df)

# Add title and labels
plt.title('Scatter Plot of Execution Time vs File Name')
plt.xlabel('File Name')
plt.ylabel('Execution Time')

# Show the plot
plt.show()
```

```python
# Plot 'dataset_content_length' against 'execution_time' using Seaborn
plt.figure(figsize=(10, 6))
sns.scatterplot(x='num_formulae', y='execution_time', data=filtered_df)

# Add title and labels
plt.title('Scatter Plot of Execution Time vs Length of Dataset Content')
plt.xlabel('Complexity of KB')
plt.ylabel('Execution Time')

# Show the plot
plt.show()
```

```python
# Plot 'dataset_content_length' against 'execution_time' using Seaborn
plt.figure(figsize=(10, 6))
sns.scatterplot(x='num_formulae', y='pruned_branches', data=filtered_df)

# Add title and labels
plt.title('Scatter Plot of Execution Time vs Length of Dataset Content')
plt.xlabel('Complexity of KB')
plt.ylabel('pruned_branches')

# Show the plot
plt.show()
```

```python
# Filter out rows where 'file_name' contains '3_5_15' and 'strategy_param' is greater than or equal to 1
filtered_df = df[(df['file_name'].str.contains('3_5_15')) & (df['strategy_param'] >= 1)]

# Plot 'execution_time' against 'pruned_branches' using Seaborn with different colors for each 'strategy_param'
plt.figure(figsize=(10, 6))
sns.scatterplot(x='execution_time', y='pruned_branches', hue='strategy_param', data=filtered_df, palette='deep')

# Add title and labels
plt.title('Scatter Plot of Execution Time vs Pruned Branches with Different Strategies')
plt.xlabel('Execution Time')
plt.ylabel('Pruned Branches')

# Show the plot
plt.show()
```
